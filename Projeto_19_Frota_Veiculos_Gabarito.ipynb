
# Projeto 19 — Evolução da Frota de Veículos por Tipo (2010–2024)

**Objetivo:** Analisar o crescimento da frota de veículos ao longo do tempo, identificando padrões por tipo de veículo.

---



## 1. Problema

Como a frota de veículos evoluiu ao longo do tempo e quais tipos de veículos cresceram mais?

---



## 2. KPI

- Total de veículos por tipo  
- Taxa de crescimento  

👉 Objetivo: identificar padrões de crescimento da frota


```

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_excel("Simulacao_Frota_Veiculos_2010_2024.xlsx")
df.head()
```

```

df.info()
df.isnull().sum()
```

```

df = df.rename(columns={
    "Quantidade": "Quantidade",
    "Tipo Veiculo": "Tipo"
})

df.head()
```

```

# Total por tipo
total_tipo = df.groupby("Tipo")["Quantidade"].sum().sort_values(ascending=False)
total_tipo
```

```

# Evolução temporal
total_ano = df.groupby("Ano")["Quantidade"].sum()
total_ano
```

```

# Evolução por tipo
plt.figure(figsize=(12,6))
sns.lineplot(data=df, x="Ano", y="Quantidade", hue="Tipo")
plt.title("Evolução da Frota por Tipo")
plt.show()
```

```

# Crescimento percentual
df = df.sort_values(["Tipo","Ano"])
df["Crescimento"] = df.groupby("Tipo")["Quantidade"].pct_change()*100
df.head()
```


## Interpretação

✔ Crescimento:
> Alguns tipos de veículos crescem mais rapidamente, indicando mudança de comportamento.

✔ Tendência:
> A frota tende a aumentar ao longo do tempo.
