
# Projeto 17 — Distribuição de Emendas Parlamentares por Região (2015–2024)

**Objetivo:** Analisar a distribuição de recursos de emendas parlamentares entre regiões e identificar concentração e desigualdade.

---



## 1. Problema

Como os recursos de emendas parlamentares estão distribuídos entre regiões e existe concentração em determinadas áreas?

---



## 2. KPI

- Valor total de emendas  
- Valor médio por região  

👉 Objetivo: identificar concentração de recursos.


```

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_excel("Simulacao_Emendas_Parlamentares_2015_2024.xlsx")
df.head()
```

```

df.info()
df.isnull().sum()
```

```

df = df.rename(columns={
    "Valor Emenda": "Valor",
    "Regiao": "Regiao"
})

df.head()
```

```

# Total por região
total_regiao = df.groupby("Regiao")["Valor"].sum().sort_values(ascending=False)
total_regiao
```

```

# Evolução temporal
total_ano = df.groupby("Ano")["Valor"].sum()
total_ano
```

```

# Gráfico regiões
total_regiao.plot(kind="bar", title="Distribuição de Emendas por Região")
plt.show()
```

```

# Gráfico temporal
total_ano.plot(title="Evolução das Emendas ao Longo do Tempo")
plt.show()
```


## Interpretação

✔ Concentração:
> Algumas regiões recebem mais recursos, indicando possível desigualdade.

✔ Distribuição equilibrada:
> Recursos distribuídos de forma homogênea.
