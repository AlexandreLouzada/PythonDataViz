
# Projeto 8 — Análise da Qualidade do Ar nas Capitais Brasileiras (2015–2023)

## Objetivo
Analisar a qualidade do ar nas capitais brasileiras ao longo do tempo, identificando cidades mais críticas e poluentes mais relevantes.

---



## 1. Problema

Como a qualidade do ar evoluiu nas capitais brasileiras entre 2015 e 2023 e quais cidades apresentam os piores índices?

---



## 2. KPI

### KPI principal:
- Índice de poluição do ar

### KPI complementar:
- Média por cidade
- Média por poluente

👉 Objetivo: identificar cidades mais críticas e padrões ao longo do tempo.


```

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_excel("Simulacao_Qualidade_Ar_Capitais_2015_2023.xlsx")
df.head()
```

```

df.info()
df.isnull().sum()
```

```

# Ajuste de colunas (exemplo)
df = df.rename(columns={
    "Cidade": "Cidade",
    "Poluente": "Poluente",
    "Indice Poluicao": "Indice"
})

df.head()
```

```

# Média por cidade
media_cidade = df.groupby("Cidade")["Indice"].mean().sort_values(ascending=False)
media_cidade
```

```

# Média por ano
media_ano = df.groupby("Ano")["Indice"].mean()
media_ano
```

```

# Média por poluente
media_poluente = df.groupby("Poluente")["Indice"].mean().sort_values(ascending=False)
media_poluente
```

```

# Gráfico evolução temporal
media_ano.plot(title="Evolução da Qualidade do Ar")
plt.show()
```

```

# Ranking cidades
media_cidade.head(10).plot(kind="bar", title="Cidades mais poluídas")
plt.show()
```

```

# Heatmap cidade x poluente
pivot = df.pivot_table(index="Cidade", columns="Poluente", values="Indice", aggfunc="mean")

sns.heatmap(pivot, cmap="Reds")
plt.title("Poluição por Cidade e Poluente")
plt.show()
```


## 6. Interpretação

### Exemplo esperado:

> Algumas cidades apresentam níveis consistentemente mais altos de poluição, indicando problemas estruturais como trânsito intenso ou atividade industrial.

> Certos poluentes são mais predominantes, sugerindo fontes específicas de emissão.
