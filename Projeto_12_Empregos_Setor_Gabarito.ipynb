## 1. Introdução

A análise do mercado de trabalho é essencial para compreender o comportamento da economia e identificar setores com maior geração ou perda de empregos. Ao investigar admissões, desligamentos e saldo de empregos, é possível observar períodos de crise, retomada e mudanças no dinamismo econômico.

Neste projeto, vamos analisar a geração de empregos por setor econômico entre 2015 e 2024.



## 2. Definição do problema

**Pergunta de análise:**  
Como evoluiu a geração de empregos por setor econômico entre 2015 e 2024 e quais setores apresentaram maior crescimento ou retração?

**Por que essa pergunta é relevante?**  
Porque ajuda a identificar quais setores impulsionam o mercado de trabalho e quais são mais vulneráveis a oscilações econômicas.

**Qual decisão poderia ser tomada com base nessa análise?**  
Direcionar políticas públicas de emprego, qualificação profissional e incentivos econômicos para setores estratégicos ou mais afetados.



## 3. Definição do KPI

### KPI principal
**Saldo de empregos**

### Fórmula
\[
	ext{Saldo de Empregos} = 	ext{Admissões} - 	ext{Desligamentos}
\]

### Justificativa
Esse indicador permite identificar se o setor gerou ou perdeu empregos em determinado período.

### KPI complementar
- Total de admissões por setor
- Total de desligamentos por setor
- Variação percentual do saldo ao longo do tempo


```

# 4. Aquisição dos dados

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["figure.figsize"] = (10, 6)
sns.set_theme(style="whitegrid")

# Leitura da base
df = pd.read_excel("Simulacao_Empregos_Setor_Regiao_2015_2024.xlsx")

# Visualização inicial
df.head()
```


### Descrição das variáveis principais
- **Ano**: ano de referência
- **Setor**: setor econômico analisado
- **Admissões**: quantidade de contratações
- **Desligamentos**: quantidade de demissões
- **Região** (se disponível): localização geográfica do registro


```

# 5. Limpeza e preparação dos dados

print("Dimensões da base:", df.shape)
print("\nTipos de dados:")
print(df.dtypes)

print("\nValores ausentes por coluna:")
print(df.isnull().sum())

print("\nQuantidade de linhas duplicadas:", df.duplicated().sum())
```

```

# Padronização de nomes de colunas (ajuste defensivo)
rename_map = {
    "Admissoes": "Admissoes",
    "Admissões": "Admissoes",
    "Desligamentos": "Desligamentos",
    "Setor Econômico": "Setor",
    "Setor_Economico": "Setor"
}
df = df.rename(columns={k: v for k, v in rename_map.items() if k in df.columns})

# Criação do KPI principal
df["Saldo_Empregos"] = df["Admissoes"] - df["Desligamentos"]

df.head()
```


### Comentário sobre a preparação
Nesta etapa:
- verificamos a qualidade da base;
- ajustamos nomes de colunas, quando necessário;
- criamos a variável **Saldo_Empregos**, que é o KPI principal do projeto.


```

# 6. Análise exploratória

# Estatísticas descritivas do saldo
df["Saldo_Empregos"].describe()
```

```

# Saldo total por ano
saldo_por_ano = df.groupby("Ano")["Saldo_Empregos"].sum().reset_index()
saldo_por_ano
```

```

# Saldo total por setor
saldo_por_setor = (
    df.groupby("Setor")["Saldo_Empregos"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)
saldo_por_setor
```

```

# Tabela dinâmica: saldo por ano e por setor
saldo_setor_ano = df.pivot_table(
    index="Ano",
    columns="Setor",
    values="Saldo_Empregos",
    aggfunc="sum"
)

saldo_setor_ano
```

```

# Variação percentual do saldo por setor ao longo do tempo
df = df.sort_values(["Setor", "Ano"])
df["Variacao_Percentual"] = df.groupby("Setor")["Saldo_Empregos"].pct_change() * 100

df.head(10)
```


## 7. Visualizações

Os gráficos abaixo ajudam a responder:
- quais setores mais geram empregos;
- quais setores perderam empregos;
- em quais anos ocorreram períodos de crise ou recuperação.


```

# Gráfico 1: saldo total de empregos ao longo do tempo
plt.figure(figsize=(10, 6))
sns.lineplot(data=saldo_por_ano, x="Ano", y="Saldo_Empregos", marker="o")
plt.title("Evolução do Saldo Total de Empregos (2015–2024)")
plt.xlabel("Ano")
plt.ylabel("Saldo de Empregos")
plt.xticks(saldo_por_ano["Ano"], rotation=45)
plt.show()
```

```

# Gráfico 2: saldo total por setor
plt.figure(figsize=(8, 5))
sns.barplot(data=saldo_por_setor, x="Setor", y="Saldo_Empregos")
plt.title("Saldo Total de Empregos por Setor")
plt.xlabel("Setor")
plt.ylabel("Saldo de Empregos")
plt.xticks(rotation=45)
plt.show()
```

```

# Gráfico 3: evolução do saldo por setor
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x="Ano", y="Saldo_Empregos", hue="Setor", marker="o")
plt.title("Evolução do Saldo de Empregos por Setor")
plt.xlabel("Ano")
plt.ylabel("Saldo de Empregos")
plt.xticks(sorted(df["Ano"].unique()), rotation=45)
plt.legend(title="Setor")
plt.show()
```

```

# Gráfico 4: distribuição dos saldos
plt.figure(figsize=(8, 5))
sns.histplot(df["Saldo_Empregos"], bins=10, kde=True)
plt.title("Distribuição do Saldo de Empregos")
plt.xlabel("Saldo de Empregos")
plt.ylabel("Frequência")
plt.show()
```


## 8. Insights e interpretação

Nesta etapa, o foco é explicar o comportamento dos setores, e não apenas mostrar gráficos.


```

# Setor com maior saldo acumulado
setor_maior_saldo = saldo_por_setor.loc[saldo_por_setor["Saldo_Empregos"].idxmax()]
setor_maior_saldo
```

```

# Ano com pior saldo total
ano_pior_saldo = saldo_por_ano.loc[saldo_por_ano["Saldo_Empregos"].idxmin()]
ano_pior_saldo
```


### Modelo de interpretação esperada

**Insight 1 — Setores mais dinâmicos**  
A análise do saldo de empregos mostra quais setores mais contribuíram para a geração de postos de trabalho ao longo do período analisado. Setores com saldo positivo recorrente tendem a demonstrar maior dinamismo econômico.

**Insight 2 — Períodos de retração e recuperação**  
A série temporal permite identificar anos de pior desempenho, que podem estar associados a crises econômicas, mudanças no mercado ou choques externos. Da mesma forma, anos de recuperação indicam retomada da atividade econômica.

**Insight 3 — Diferenças entre setores**  
Nem todos os setores reagem da mesma forma às mudanças econômicas. Alguns mantêm crescimento mais estável, enquanto outros apresentam maior sensibilidade, com oscilações mais intensas.

**Insight 4 — Limitações da análise**  
A base permite observar o comportamento de admissões, desligamentos e saldo, mas não explica sozinha os fatores estruturais por trás dessas mudanças. Uma análise mais ampla exigiria dados de inflação, PIB, consumo e investimento.



## 9. Conclusão

Com base na análise realizada, é possível identificar quais setores econômicos mais geraram empregos, quais apresentaram maior retração e em quais períodos ocorreram oscilações importantes no mercado de trabalho.

Esses resultados podem apoiar decisões de políticas públicas, ações de qualificação profissional e estratégias de desenvolvimento econômico setorial.

A geração de empregos variou significativamente entre os setores e ao longo do tempo, com setores mais dinâmicos e outros mais vulneráveis a períodos de crise ou retração.
