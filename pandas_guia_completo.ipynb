### Seleção e Fatiamento

#### `.loc[]`

- **Explicação:** Seleciona dados baseando-se em rótulos de linhas e nomes das colunas. Aceita condições booleanas.
- **Sintaxe:** `df.loc[linhas_rótulo_ou_condição, colunas_rótulo]`

```python
df.loc[df['valor'] > 100, ['cliente_id', 'valor']]
```

#### `.iloc[]`

- **Explicação:** Seleciona dados através de seus índices inteiros posicionais (base zero).
- **Sintaxe:** `df.iloc[inicio_linha:fim_linha, inicio_coluna:fim_coluna]`

```python
df.iloc[0:5, 0:2]  # Primeiras 5 linhas e primeiras 2 colunas
```

#### `.isin()`

- **Explicação:** Filtra registros cujos valores de uma coluna estejam presentes em uma lista especificada.
- **Sintaxe:** `df['coluna'].isin([valor1, valor2])`

```python
df[df['estado'].isin(['SP', 'RJ'])]
```

### Limpeza e Tratamento

#### `.dropna()`

- **Explicação:** Remove linhas ou colunas que contenham valores nulos (`NaN`).
- **Sintaxe:** `df.dropna(subset=['coluna_chave'], how='any', inplace=False)`

```python
df_limpo = df.dropna(subset=['valor_venda'])  # Remove linhas com valor_venda nulo
```

#### `.fillna()`

- **Explicação:** Preenche valores nulos (`NaN`) por um valor fixo ou calculado (como média ou mediana).
- **Sintaxe:** `df['coluna'].fillna(value)`

```python
df['valor_venda'] = df['valor_venda'].fillna(df['valor_venda'].median())
```

#### `.drop()`

- **Explicação:** Elimina colunas ou linhas especificadas do DataFrame.
- **Sintaxe:** `df.drop(columns=['col1', 'col2'])`

```python
df_reduzido = df.drop(columns=['rascunho', 'temp'])
```

#### `.rename()`

- **Explicação:** Renomeia os rótulos de colunas ou índices.
- **Sintaxe:** `df.rename(columns={'nome_antigo': 'nome_novo'})`

```python
df = df.rename(columns={'val': 'valor_venda', 'cli': 'cliente_id'})
```

### Agrupamento e Reestruturação

#### `.groupby()` e `.agg()`

- **Explicação:** Agrupa dados por uma ou mais categorias e aplica funções de agregação (soma, média, contagem) sobre as variáveis.
- **Sintaxe:** `df.groupby('coluna_grupo').agg({'coluna_valor': 'função'})`

```python
df.groupby('regiao').agg({'valor_venda': ['sum', 'mean']})
```

#### `.transform()`

- **Explicação:** Aplica uma função de grupo e retorna um objeto com a mesma quantidade de linhas do DataFrame original (ideal para calcular percentuais do total do grupo).
- **Sintaxe:** `df.groupby('grupo')['coluna'].transform('funcao')`

```python
df['media_categoria'] = df.groupby('categoria')['valor'].transform('mean')
```

#### `pd.pivot_table()`

- **Explicação:** Constrói uma tabela dinâmica relacionando índices, colunas e uma função agregadora sobre valores.
- **Sintaxe:** `pd.pivot_table(df, values='v', index='i', columns='c', aggfunc='sum')`

```python
pd.pivot_table(df, values='valor', index='categoria', columns='estado', aggfunc='sum', fill_value=0)
```

#### `pd.melt()`

- **Explicação:** Transforma o DataFrame do formato *wide* (largo) para o formato *long* (comprido), unificando várias colunas em pares de variável-valor.
- **Sintaxe:** `pd.melt(df, id_vars=['id'], value_vars=['col1', 'col2'], var_name='var', value_name='val')`

```python
pd.melt(df, id_vars=['cliente_id'], value_vars=['jan', 'fev'], var_name='mes', value_name='venda')
```

#### `.unstack()`

- **Explicação:** Desempilha um nível dos índices rotulados (`MultiIndex`) para as colunas do DataFrame.
- **Sintaxe:** `df.unstack(level=-1)`

```python
df.groupby(['regiao', 'categoria'])['valor'].sum().unstack()
```

### Séries Temporais

#### `pd.to_datetime()`

- **Explicação:** Converte textos ou números em objetos no formato de data e hora nativo do pandas (`datetime64`).
- **Sintaxe:** `pd.to_datetime(arg, format=None)`

```python
df['data_hora'] = pd.to_datetime(df['data_hora'], format='%Y-%m-%d %H:%M:%S')
```

#### `.dt` (Acessor Temporal)

- **Explicação:** Dá acesso a propriedades de data como ano, mês, dia, dia da semana, hora, etc.
- **Sintaxe:** `df['coluna_datetime'].dt.propriedade`

```python
df['ano'] = df['data_hora'].dt.year
df['dia_semana'] = df['data_hora'].dt.day_name()
```

#### `.resample()`

- **Explicação:** Agrupa dados temporais por frequências customizadas (diária, mensal, trimestral).
- **Sintaxe:** `df.set_index('data').resample('rule').sum()`

```python
df.set_index('data_hora').resample('ME')['valor'].sum()  # Agrupamento ao fim de cada mês
```

#### `.shift()`

- **Explicação:** Desloca os valores de uma coluna por um número determinado de períodos (útil para calcular variação percentual em relação ao mês anterior).
- **Sintaxe:** `df['coluna'].shift(periods=1)`

```python
df['venda_mes_anterior'] = df['valor_venda'].shift(1)
```

#### `.str.contains()` / `.str.replace()` / `.str.extract()`

- **Explicação:** Aplica operações e expressões regulares diretamente em colunas de texto.

### Vetorização de Condicionais

#### `np.where()`

- **Explicação:** Operação condicional vetorizada similar ao comando IF do Excel.
- **Sintaxe:** `np.where(condicao, valor_se_verdadeiro, valor_se_falso)`

```python
import numpy as np

df['status_valor'] = np.where(df['valor'] > 500, 'Alto', 'Baixo')
```

#### `np.select()`

- **Explicação:** Permite criar colunas condicionais com múltiplas regras e resultados em código compilado C.
- **Sintaxe:** `np.select(lista_condicoes, lista_escolhas, default)`

```python
condicoes = [df['valor'] >= 1000, df['valor'] >= 500]
escolhas = ['Classe A', 'Classe B']

df['categoria_cliente'] = np.select(condicoes, escolhas, default='Classe C')
```

#### `.query()` e `.eval()`

- **Explicação:** Executa filtragens e expressões matemáticas diretamente com sintaxe otimizada baseada em C e NumExpr.
- **Sintaxe:** `df.query('expressão_string')`

```python
df_filtrado = df.query("valor > 500 and status == 'Concluído'")
```

### Pipelines e Processamento de Dados Estruturados

#### `.assign()`

- **Explicação:** Cria ou altera colunas em um DataFrame retornando um novo objeto, permitindo o encadeamento de métodos.
- **Sintaxe:** `df.assign(nova_coluna=expressao)`

```python
df_novo = df.assign(imposto=lambda x: x['valor'] * 0.1)
```

#### `.pipe()`

- **Explicação:** Aplica funções customizadas encadeadas para formar um pipeline de transformação contínuo e modular.
- **Sintaxe:** `df.pipe(funcao_customizada)`

```python
def remover_nulos(df):
    return df.dropna(subset=['id'])

df_final = df.pipe(remover_nulos)
```

#### `pd.json_normalize()`

- **Explicação:** Transforma dados no formato JSON hierárquico/aninhado em uma tabela plana regular.
- **Sintaxe:** `pd.json_normalize(dados_json)`

```python
import json

df_flat = pd.json_normalize(df['detalhes_json'].apply(json.loads))
```

#### `.explode()`

- **Explicação:** Transforma cada elemento de uma lista presente em uma célula em uma nova linha individual no DataFrame.
- **Sintaxe:** `df.explode('coluna_com_lista')`

```python
df_expandido = df.explode('tags_produtos')
```
