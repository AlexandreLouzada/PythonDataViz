
# Projeto 4 — Evasão Escolar no Ensino Médio por Estado (2010–2020)

## Objetivo
Analisar a evolução da evasão escolar no ensino médio ao longo do tempo e identificar quais estados apresentam maiores taxas.

---



## 1. Problema

Como a taxa de evasão escolar no ensino médio evoluiu entre 2010 e 2020 e quais estados apresentam maior nível de evasão?

---



## 2. KPI

### KPI principal:
- Taxa de evasão escolar (%)

### Justificativa:
Permite comparar proporcionalmente os estados e identificar desigualdades educacionais.


```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_excel("Simulacao_Evasao_Escolar_Ensino_Medio_2010_2020.xlsx")
df.head()
```

```python

df.info()
df.isnull().sum()
```

```python

# Ajuste de colunas
df = df.rename(columns={"Taxa de Evasão (%)": "Taxa_Evasao"})
df.head()
```

```python

# Evolução temporal
evasao_ano = df.groupby("Ano")["Taxa_Evasao"].mean()
evasao_ano
```

```python

# Ranking de estados
evasao_estado = df.groupby("Estado")["Taxa_Evasao"].mean().sort_values(ascending=False)
evasao_estado
```

```python

# Evolução por estado
plt.figure(figsize=(12,6))
sns.lineplot(data=df, x="Ano", y="Taxa_Evasao", hue="Estado")
plt.title("Evolução da Evasão Escolar por Estado")
plt.show()
```

```python

# Identificação de extremos
estado_max = evasao_estado.idxmax()
estado_min = evasao_estado.idxmin()

estado_max, estado_min
```


## 6. Interpretação

Exemplo esperado:

> Estados com maiores taxas de evasão indicam possíveis problemas estruturais, como baixa qualidade do ensino, desigualdade social ou falta de políticas públicas eficazes.

> A evolução ao longo do tempo pode indicar melhora ou agravamento da situação educacional.
