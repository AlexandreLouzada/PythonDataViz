## Configuração do Ambiente

Execute a célula abaixo antes dos exemplos: importa as bibliotecas, carrega o `vendas.csv` (mesmo arquivo do tutorial de pandas) e prepara os dados de agregação usados nos gráficos.

```python
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# DataFrame de vendas (mesmo arquivo do tutorial de pandas)
df = pd.read_csv('vendas.csv', sep=';', encoding='utf-8')
df['data_hora'] = pd.to_datetime(df['data_hora'], format='%Y-%m-%d %H:%M:%S')

# Agregações prontas para os gráficos
vendas_mensal = df.set_index('data_hora')['valor'].resample('ME').sum()
vendas_categoria = df.groupby('categoria')['valor'].sum().sort_values()
vendas_regiao = df.groupby('regiao')['valor'].sum()

print('Dados carregados:', df.shape)
```

# MÓDULO 1 — Nível Básico
## Criação dos Gráficos Essenciais

---
## 1.1 Construindo Gráficos

### `plt.plot()` — Gráfico de Linha

Conecta pontos com linhas. Ideal para **séries temporais** e tendências.

**Sintaxe:** `plt.plot(x, y, format_string)`

```python
plt.figure(figsize=(8, 4))
plt.plot(vendas_mensal.index, vendas_mensal.values, marker='o', linestyle='-', color='steelblue')
plt.title('Vendas Mensais')
plt.xlabel('Mês')
plt.ylabel('Valor (R$)')
plt.grid(True)
plt.show()
```

### `plt.scatter()` — Gráfico de Dispersão

Plota pontos individuais. Ideal para ver **relações e correlações** entre duas variáveis.

**Sintaxe:** `plt.scatter(x, y, s=tamanho, c=cor, alpha=transparência)`

```python
plt.figure(figsize=(6, 5))
plt.scatter(df['valor'], df['valor_venda'], alpha=0.6, c='darkorange')
plt.title('Valor x Valor de Venda')
plt.xlabel('Valor (R$)')
plt.ylabel('Valor de Venda (R$)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### `plt.bar()` — Gráfico de Barras

Compara categorias por meio de barras. Use `barh` para barras horizontais.

**Sintaxe:** `plt.bar(labels, valores, color=cor)`

```python
plt.figure(figsize=(8, 4))
plt.bar(vendas_categoria.index, vendas_categoria.values, color='seagreen')
plt.title('Vendas por Categoria')
plt.ylabel('Valor (R$)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### `plt.hist()` — Histograma

Distribui os dados em intervalos (bins) mostrando a **frequência** dos valores.

**Sintaxe:** `plt.hist(dados, bins=n, edgecolor=cor)`

```python
plt.figure(figsize=(6, 4))
plt.hist(df['valor'], bins=10, edgecolor='black', color='mediumpurple')
plt.title('Distribuição dos Valores de Venda')
plt.xlabel('Valor (R$)')
plt.ylabel('Frequência')
plt.show()
```

### `plt.pie()` — Gráfico de Pizza

Mostra a **proporção** de cada categoria em relação ao total.

**Sintaxe:** `plt.pie(valores, labels=rótulos, autopct='%1.1f%%')`

```python
plt.figure(figsize=(6, 6))
plt.pie(vendas_regiao.values, labels=vendas_regiao.index, autopct='%1.1f%%', startangle=90)
plt.title('Participação por Região')
plt.show()
```

### `plt.boxplot()` — Diagrama de Caixa

Resume a distribuição com **quartis**, mediana e valores atípicos (*outliers*). Ideal para comparar grupos.

**Sintaxe:** `plt.boxplot([grupo1, grupo2], labels=[...])`

```python
categorias = df['categoria'].unique()
dados = [df[df['categoria'] == c]['valor'].dropna() for c in categorias]

plt.figure(figsize=(8, 4))
plt.boxplot(dados, tick_labels=categorias)
plt.title('Distribuição de Valores por Categoria')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

---
## 1.2 Personalização de Gráficos

### Título, Eixos, Limites, Grade e Legenda

Funções usadas em **quase todo** gráfico para dar contexto e clareza.

**Sintaxe:** `plt.title()`, `plt.xlabel()`, `plt.ylabel()`, `plt.xlim()`, `plt.ylim()`, `plt.grid()`, `plt.legend()`, `plt.text()`

```python
x = np.arange(1, 11)
y = x ** 2

plt.figure(figsize=(7, 4))
plt.plot(x, y, label='y = x²')
plt.plot(x, y * 0.5, label='y = x²/2', linestyle='--')

plt.title('Curvas Quadráticas')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.xlim(0, 12)
plt.ylim(0, 120)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(loc='upper left')
plt.show()
```

### Cores, Estilos de Linha e Marcadores

Controle fino do visual com `color` (nomes, hex, RGB), `linestyle` e `marker`.

**Sintaxe:** `plt.plot(x, y, color='#hex', linestyle='--', marker='o', linewidth=2)`

```python
x = np.linspace(0, 2 * np.pi, 100)

plt.figure(figsize=(8, 4))
plt.plot(x, np.sin(x), color='#e74c3c', linestyle='-', marker='', linewidth=2, label='sen(x)')
plt.plot(x, np.cos(x), color='#2980b9', linestyle='--', marker='', linewidth=2, label='cos(x)')
plt.axhline(0, color='black', linewidth=0.8)
plt.legend()
plt.title('Funções Trigonométricas')
plt.show()
```

### Escala Logarítmica

Útil quando os dados abrangem **muitas ordens de grandeza**.

**Sintaxe:** `plt.xscale('log')`, `plt.yscale('log')`

```python
x = np.logspace(0, 5, 60)
y = x ** 1.5

plt.figure(figsize=(7, 4))
plt.plot(x, y, color='teal')
plt.xscale('log')
plt.yscale('log')
plt.title('Reta no Gráfico Log-Log')
plt.xlabel('X (escala log)')
plt.ylabel('Y (escala log)')
plt.grid(True, which='both', alpha=0.3)
plt.show()
```

# MÓDULO 2 — Nível Intermediário
## Figuras, Eixos, Estilos e Integração com Pandas

---
## 2.1 API Orientada a Objetos (Figuras e Eixos)

### `plt.figure()` e `plt.subplots()`

A API orientada a objetos separa a **figura** (janela completa) dos **eixos** (área do gráfico), dando controle total.

**Sintaxe:** `fig, ax = plt.subplots(nrows, ncols, figsize=(larg, alt))`

```python
fig, ax = plt.subplots(figsize=(8, 4))

ax.plot(vendas_mensal.index, vendas_mensal.values, marker='s')
ax.set_title('Vendas Mensais — API Orientada a Objetos')
ax.set_xlabel('Mês')
ax.set_ylabel('Valor (R$)')
ax.grid(True, alpha=0.4)
plt.show()
```

### Múltiplos Gráficos (Subplots em Grade)

`plt.subplots(nrows=2, ncols=2)` cria uma grade de eixos acessados por **índices** (`ax[0,0]`, `ax[1,1]`, etc.).

**Sintaxe:** `fig, axes = plt.subplots(nrows, ncols)`

```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 7))

axes[0, 0].bar(df['estado'].value_counts().index, df['estado'].value_counts().values, color='coral')
axes[0, 0].set_title('Contagem por Estado')

axes[0, 1].hist(df['valor'].dropna(), bins=8, color='mediumseagreen', edgecolor='white')
axes[0, 1].set_title('Histograma de Valores')

axes[1, 0].scatter(df['valor'], df['valor_venda'], alpha=0.6)
axes[1, 0].set_title('Dispersão Valor x Venda')

axes[1, 1].plot(vendas_mensal.index, vendas_mensal.values, color='rebeccapurple')
axes[1, 1].set_title('Série Mensal')

plt.tight_layout()
plt.show()
```

### Compartilhamento de Eixos (`sharex` / `sharey`)

Alinha as escalas entre subplots, facilitando a **comparação visual**. `sharex` compartilha o eixo X e `sharey` o Y.

**Sintaxe:** `plt.subplots(nrows=2, sharex=True)`

```python
mensal_agg = df.set_index('data_hora').resample('ME').agg({'valor': 'sum', 'cliente_id': 'count'})

fig, (ax1, ax2) = plt.subplots(nrows=2, ncols=1, figsize=(8, 5), sharex=True)

ax1.plot(mensal_agg.index, mensal_agg['valor'], color='steelblue')
ax1.set_ylabel('Soma (R$)')
ax1.set_title('Vendas Mensais — Soma e Contagem')

ax2.bar(mensal_agg.index, mensal_agg['cliente_id'], color='tomato')
ax2.set_xlabel('Mês')
ax2.set_ylabel('Quantidade')

plt.tight_layout()
plt.show()
```

## 2.2 Estilos e Exportação

### `plt.style.use()` — Tema Pronto

Aplica **temas pré-configurados** (paleta, grades, fundo) com uma única linha.

**Sintaxe:** `plt.style.use('nome_do_estilo')` — opções: `ggplot`, `fivethirtyeight`, `seaborn-v0_8`, `bmh`, `grayscale`

```python
plt.style.use('ggplot')

plt.figure(figsize=(8, 4))
plt.bar(vendas_categoria.index, vendas_categoria.values)
plt.title('Vendas por Categoria — Estilo ggplot')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.style.use('default')  # volta ao padrão
print(plt.style.available)
```

### `plt.savefig()` — Salvar em Arquivo

Exporta a figura atual em PNG, PDF, SVG etc. `dpi` controla a resolução e `bbox_inches='tight'` ajusta as bordas.

**Sintaxe:** `plt.savefig('arquivo.png', dpi=300, bbox_inches='tight')`

```python
plt.figure(figsize=(6, 4))
plt.plot(vendas_mensal.index, vendas_mensal.values, marker='o')
plt.title('Vendas Mensais')
plt.savefig('vendas_mensais.png', dpi=200, bbox_inches='tight')
print('Arquivo salvo: vendas_mensais.png')
```

---
## 2.3 Integração com Pandas

### `df.plot()` — Plotagem Direta do DataFrame

O pandas **encapsula o matplotlib**: cada coluna vira uma série e o índice vira o eixo X.

**Sintaxe:** `df.plot(x, y, kind='linha')` — kinds: `line`, `bar`, `barh`, `hist`, `box`, `scatter`, `pie`

```python
vendas_categoria.plot(kind='bar', title='Vendas por Categoria (df.plot)',
                      xlabel='Categoria', ylabel='Valor (R$)',
                      figsize=(8, 4), rot=45)
plt.tight_layout()
plt.show()
```

### Múltiplas Séries no Mesmo Gráfico

Selecionando **várias colunas**, o pandas plota cada uma delas como uma linha/barra com legenda automática.

**Sintaxe:** `df[['col1', 'col2']].plot()`

```python
mensal = df.set_index('data_hora')[['valor', 'valor_venda']].resample('ME').sum()

mensal.plot(figsize=(8, 4), marker='o', title='Valor x Valor de Venda — Mensal')
plt.ylabel('R$')
plt.grid(True, alpha=0.4)
plt.show()
```

---
## 2.4 Anotações e Eixos Duplos

### `ax.annotate()` / `ax.text()` — Anotações

Destaca pontos específicos ou adiciona rótulos dentro do gráfico.

**Sintaxe:** `ax.annotate(texto, xy=(x, y), xytext=(x, y), arrowprops=dict(arrowstyle='->'))`

```python
fig, ax = plt.subplots(figsize=(8, 4))

ax.plot(vendas_mensal.index, vendas_mensal.values, marker='o', color='steelblue')
ax.set_title('Vendas Mensais com Destaques')
ax.set_ylabel('Valor (R$)')

pico = vendas_mensal.idxmax()
ax.annotate(f'Pico: {vendas_mensal.max():,.0f}',
            xy=(pico, vendas_mensal.max()), xytext=(pico, vendas_mensal.max() * 0.9),
            arrowprops=dict(arrowstyle='->', color='red'), color='red')

ax.text(0.02, 0.95, 'Série 2024', transform=ax.transAxes,
        fontsize=10, va='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.6))
plt.show()
```

### `ax.twinx()` — Dois Eixos Y

Sobrepoem duas séries com **escalas diferentes** no mesmo gráfico (uma à esquerda, outra à direita).

**Sintaxe:** `ax2 = ax.twinx()`

```python
fig, ax = plt.subplots(figsize=(8, 4))

mensal = df.set_index('data_hora').resample('ME').agg({'valor': 'sum', 'cliente_id': 'count'})

ax.bar(mensal.index, mensal['valor'], color='lightcoral', alpha=0.8, label='Valor (R$)')
ax.set_ylabel('Valor Total (R$)', color='darkred')

ax2 = ax.twinx()
ax2.plot(mensal.index, mensal['cliente_id'], color='darkblue', marker='o', label='Qtd. vendas')
ax2.set_ylabel('Quantidade de Vendas', color='darkblue')

ax.set_title('Valor Total e Quantidade por Mês — Eixos Duplos')
plt.show()
```

# MÓDULO 3 — Nível Avançado

## 3.1 Layouts Complexos com `GridSpec`

### `GridSpec`

Cria grades de subplots com **tamanhos variáveis**, permitindo que um gráfico ocupe várias células.

**Sintaxe:** `GridSpec(nrows, ncols, width_ratios=[...], height_ratios=[...])` e `fig.add_subplot(gs[fatias])`

```python
from matplotlib.gridspec import GridSpec

fig = plt.figure(figsize=(10, 6))
gs = GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[2, 1])

ax_larga = fig.add_subplot(gs[0, 0])
ax_direita = fig.add_subplot(gs[:, 1])
ax_inferior = fig.add_subplot(gs[1, 0])

ax_larga.plot(vendas_mensal.index, vendas_mensal.values, marker='o', color='steelblue')
ax_larga.set_title('Série Mensal')

ax_direita.pie(vendas_regiao.values, labels=vendas_regiao.index, autopct='%1.0f%%')
ax_direita.set_title('Regiões')

ax_inferior.bar(vendas_categoria.index, vendas_categoria.values, color='seagreen')
ax_inferior.set_title('Categorias')
ax_inferior.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

## 3.2 Heatmaps e `imshow`

### `imshow()` — Matriz de Intensidade

Renderiza uma matriz como uma **grade de cores**. Use `cmap` para escolher a paleta e `colorbar` para a escala.

**Sintaxe:** `plt.imshow(matriz, cmap='viridis')` + `plt.colorbar()`

```python
pivot = pd.pivot_table(df, values='valor', index='categoria', columns='estado', aggfunc='sum', fill_value=0)

plt.figure(figsize=(8, 5))
plt.imshow(pivot.values, cmap='YlGnBu', aspect='auto')
plt.colorbar(label='Valor (R$)')
plt.xticks(range(len(pivot.columns)), pivot.columns)
plt.yticks(range(len(pivot.index)), pivot.index)
plt.title('Heatmap: Valor por Categoria x Estado')
plt.tight_layout()
plt.show()
```

### `plt.contourf()` — Contornos Preenchidos

Mostra **superfícies contínuas** de dados 2D com curvas de nível preenchidas.

**Sintaxe:** `plt.contourf(X, Y, Z, levels=n, cmap='paleta')`

```python
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y)

plt.figure(figsize=(6, 5))
plt.contourf(X, Y, Z, levels=20, cmap='coolwarm')
plt.colorbar()
plt.title('Contornos: sen(x) * cos(y)')
plt.show()
```

---
## 3.3 Gráficos Tridimensionais

### Plotagem 3D

Baseado no `mpl_toolkits.mplot3d`. Cria-se um eixo 3D e usa-se `plot_surface`, `scatter`, `plot` de forma similar ao 2D.

**Sintaxe:** `fig = plt.figure()` + `ax = fig.add_subplot(projection='3d')`

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(7, 6))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)
ax.set_title('Superfície 3D — sen(x) * cos(y)')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
plt.show()
```

### Animações com `FuncAnimation`

Gera um objeto animado atualizando os dados da linha a cada *frame*. Em notebooks, exiba com `plt.show()` ou salve em GIF/vídeo.

**Sintaxe:** `FuncAnimation(fig, funcao_de_update, frames=n, interval=ms)`

```python
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

fig, ax = plt.subplots(figsize=(6, 4))
x = np.linspace(0, 2 * np.pi, 100)
linha, = ax.plot([], [])
ax.set_xlim(0, 2 * np.pi)
ax.set_ylim(-1.2, 1.2)
ax.set_title('Onda Senoidal Animada')

def update(fase):
    linha.set_data(x, np.sin(x + fase / 10))
    return linha,

anim = FuncAnimation(fig, update, frames=100, interval=40, blit=True)

# Para exibir a animação no notebook, use: HTML(anim.to_jshtml())
print('Animação criada. Para exibir, execute: HTML(anim.to_jshtml())')
```

## 3.4 Configuração Global com `rcParams`

### `plt.rcParams`

Define valores **padrão globais** (tamanho da letra, resolução, cor da grade) para todos os gráficos da sessão.

**Sintaxe:** `plt.rcParams['chave'] = valor` — chaves comuns: `figure.figsize`, `font.size`, `axes.grid`, `savefig.dpi`

```python
print('Antes de alterar:')
print('  figure.figsize =', plt.rcParams['figure.figsize'])
print('  font.size =', plt.rcParams['font.size'])

plt.rcParams['figure.figsize'] = (9, 5)
plt.rcParams['font.size'] = 11
plt.rcParams['axes.grid'] = True
plt.rcParams['axes.grid.alpha'] = 0.3

plt.plot(vendas_mensal.index, vendas_mensal.values, marker='o')
plt.title('Gráfico com padrões globais aplicados')
plt.show()

# Restaura os padrões
plt.rcdefaults()
```
