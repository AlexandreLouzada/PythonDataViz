
# Projeto 10 — Roubos e Furtos de Veículos por Cidade (2018–2024)

## Objetivo
Analisar a evolução dos roubos e furtos de veículos ao longo do tempo, identificando padrões por cidade e possíveis períodos críticos.

---



## 1. Problema

Como os roubos e furtos de veículos evoluíram ao longo do tempo e quais cidades apresentam maior incidência?

---



## 2. KPI

### KPI principal:
- Total de ocorrências de roubos e furtos

### KPI complementar:
- Média por cidade
- Evolução temporal

👉 Objetivo: identificar cidades mais críticas e períodos de maior risco.


```

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_excel("Simulacao_Roubos_Furtos_Veiculos_Cidades_2018_2024.xlsx")
df.head()
```

```

df.info()
df.isnull().sum()
```

```

# Ajuste de colunas
df = df.rename(columns={
    "Ocorrencias": "Ocorrencias",
    "Cidade": "Cidade"
})

df.head()
```

```

# Total por ano
ocorrencias_ano = df.groupby("Ano")["Ocorrencias"].sum()
ocorrencias_ano
```

```

# Total por cidade
ocorrencias_cidade = df.groupby("Cidade")["Ocorrencias"].sum().sort_values(ascending=False)
ocorrencias_cidade
```

```

# Evolução temporal
plt.figure(figsize=(10,6))
sns.lineplot(data=df, x="Ano", y="Ocorrencias", hue="Cidade")
plt.title("Evolução dos Roubos e Furtos por Cidade")
plt.show()
```

```

# Ranking cidades
ocorrencias_cidade.head(10).plot(kind="bar", title="Cidades com mais ocorrências")
plt.show()
```

```

# Identificação de extremos
cidade_max = ocorrencias_cidade.idxmax()
cidade_min = ocorrencias_cidade.idxmin()

cidade_max, cidade_min
```


## 6. Interpretação

Exemplo esperado:

> Algumas cidades concentram maior número de ocorrências, indicando maior risco ou maior densidade populacional.

> A evolução temporal pode indicar aumento da criminalidade ou mudanças em políticas de segurança.

A análise permite identificar cidades mais críticas e períodos de maior incidência.
