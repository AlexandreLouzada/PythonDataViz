## 1. Introdução

Os óbitos por causas evitáveis representam um importante indicador de qualidade da atenção à saúde, do acesso a serviços e da efetividade de políticas públicas de prevenção. Analisar esse tipo de dado por faixa etária ajuda a identificar grupos mais vulneráveis e orientar ações específicas.

Neste projeto, vamos investigar a evolução dos óbitos por causas evitáveis no Brasil entre 2015 e 2024, com foco nas diferenças entre faixas etárias.



## 2. Definição do problema

**Pergunta de análise:**  
Quais faixas etárias concentram mais óbitos por causas evitáveis no período analisado e como esse comportamento evolui ao longo do tempo?

**Por que essa pergunta é relevante?**  
Porque permite identificar grupos etários mais vulneráveis, apoiando ações de prevenção, campanhas de saúde pública e alocação mais eficiente de recursos.

**Qual decisão poderia ser tomada com base nessa análise?**  
Priorizar políticas e programas de saúde voltados às faixas etárias com maior concentração de óbitos evitáveis.



## 3. Definição do KPI

### KPI principal
**Total de óbitos por causas evitáveis por faixa etária**

### Justificativa
Esse indicador permite comparar o impacto do problema entre diferentes grupos populacionais.

### Fórmula
\[
	ext{Total de Óbitos} = \sum 	ext{Óbitos por Causas Evitáveis}
\]

### KPI complementar
**Variação percentual anual dos óbitos por faixa etária**

Esse indicador ajuda a identificar grupos em crescimento, estabilidade ou redução.


```

# 4. Aquisição dos dados

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["figure.figsize"] = (10, 6)
sns.set_theme(style="whitegrid")

# Leitura da base
df = pd.read_excel("Simulacao_Obitos_Causas_Evitaveis_2015_2024.xlsx")

# Visualização inicial
df.head()
```


### Descrição das variáveis principais
- **Ano**: ano de referência dos registros
- **Faixa Etária**: grupo etário analisado
- **Óbitos por Causas Evitáveis**: quantidade de óbitos registrados


```

# 5. Limpeza e preparação dos dados

print("Dimensões da base:", df.shape)
print("\nTipos de dados:")
print(df.dtypes)

print("\nValores ausentes por coluna:")
print(df.isnull().sum())

print("\nQuantidade de linhas duplicadas:", df.duplicated().sum())
```

```

# Renomeando colunas para facilitar os comandos
df = df.rename(columns={
    "Faixa Etária": "Faixa_Etaria",
    "Óbitos por Causas Evitáveis": "Obitos_Evitaveis"
})

# Garantindo tipos corretos
df["Ano"] = df["Ano"].astype(int)
df["Faixa_Etaria"] = df["Faixa_Etaria"].astype(str)
df["Obitos_Evitaveis"] = df["Obitos_Evitaveis"].astype(int)

df.head()
```


### Observação importante sobre a base
O enunciado original menciona comparação por região, porém a base disponibilizada contém apenas:
- ano,
- faixa etária,
- quantidade de óbitos.

Portanto, este gabarito trabalha de forma completa com as variáveis disponíveis e registra essa limitação metodológica na interpretação final.


```

# 6. Análise exploratória

# Estatísticas descritivas
df["Obitos_Evitaveis"].describe()
```

```

# Total de óbitos por ano
obitos_por_ano = df.groupby("Ano")["Obitos_Evitaveis"].sum().reset_index()
obitos_por_ano
```

```

# Total de óbitos por faixa etária
obitos_por_faixa = (
    df.groupby("Faixa_Etaria")["Obitos_Evitaveis"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)
obitos_por_faixa
```

```

# Tabela dinâmica: evolução temporal por faixa etária
obitos_faixa_ano = df.pivot_table(
    index="Ano",
    columns="Faixa_Etaria",
    values="Obitos_Evitaveis",
    aggfunc="sum"
)

obitos_faixa_ano
```

```

# Cálculo da variação percentual anual por faixa etária
df = df.sort_values(["Faixa_Etaria", "Ano"])
df["Variacao_Percentual"] = df.groupby("Faixa_Etaria")["Obitos_Evitaveis"].pct_change() * 100

df.head(10)
```


## 7. Visualizações
Os gráficos abaixo ajudam a responder:
- quais faixas etárias concentram mais óbitos;
- como os óbitos evoluem ao longo do tempo;
- quais grupos merecem maior atenção.


```

# Gráfico 1: evolução total dos óbitos ao longo do tempo
plt.figure(figsize=(10, 6))
sns.lineplot(data=obitos_por_ano, x="Ano", y="Obitos_Evitaveis", marker="o")
plt.title("Evolução Total dos Óbitos por Causas Evitáveis (2015–2024)")
plt.xlabel("Ano")
plt.ylabel("Óbitos")
plt.xticks(obitos_por_ano["Ano"], rotation=45)
plt.show()
```

```

# Gráfico 2: ranking das faixas etárias
plt.figure(figsize=(10, 6))
sns.barplot(data=obitos_por_faixa, x="Faixa_Etaria", y="Obitos_Evitaveis")
plt.title("Total de Óbitos por Causas Evitáveis por Faixa Etária")
plt.xlabel("Faixa Etária")
plt.ylabel("Óbitos")
plt.xticks(rotation=45)
plt.show()
```

```

# Gráfico 3: evolução dos óbitos por faixa etária
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x="Ano", y="Obitos_Evitaveis", hue="Faixa_Etaria", marker="o")
plt.title("Evolução dos Óbitos por Causas Evitáveis por Faixa Etária")
plt.xlabel("Ano")
plt.ylabel("Óbitos")
plt.xticks(sorted(df["Ano"].unique()), rotation=45)
plt.legend(title="Faixa Etária", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```

# Gráfico 4: distribuição dos valores
plt.figure(figsize=(8, 5))
sns.histplot(df["Obitos_Evitaveis"], bins=10, kde=True)
plt.title("Distribuição dos Óbitos por Causas Evitáveis")
plt.xlabel("Óbitos")
plt.ylabel("Frequência")
plt.show()
```


## 8. Insights e interpretação
Nesta etapa, não basta descrever os gráficos. É necessário explicar o que eles indicam e quais decisões podem ser tomadas.


```

# Identificação da faixa etária com maior volume de óbitos
faixa_mais_afetada = obitos_por_faixa.loc[obitos_por_faixa["Obitos_Evitaveis"].idxmax()]
faixa_mais_afetada
```

```

# Identificação do ano com maior total de óbitos
ano_mais_critico = obitos_por_ano.loc[obitos_por_ano["Obitos_Evitaveis"].idxmax()]
ano_mais_critico
```


### Modelo de interpretação esperada

**Insight 1 — Faixas etárias mais vulneráveis**  
A análise mostra que algumas faixas etárias concentram mais óbitos por causas evitáveis do que outras. Isso pode indicar maior fragilidade biológica, menor acesso à prevenção ou maior exposição a fatores de risco específicos.

**Insight 2 — Evolução temporal**  
Ao observar a série histórica, é possível identificar se o problema se mantém estável, cresce ou apresenta oscilações importantes ao longo do tempo. Anos com maior volume de óbitos podem indicar falhas em políticas preventivas, crises sanitárias ou dificuldade de acesso ao cuidado.

**Insight 3 — Priorização de políticas públicas**  
As faixas etárias mais afetadas devem ser consideradas prioritárias em programas de saúde, campanhas educativas e ampliação de cobertura assistencial.

**Insight 4 — Limitação dos dados**  
Como a base não traz a variável região, não é possível responder de forma completa à parte regional do enunciado. Portanto, as conclusões aqui se concentram em faixa etária e tempo, respeitando a estrutura real dos dados.



## 9. Conclusão

Com base na análise, é possível identificar quais faixas etárias apresentam maior concentração de óbitos por causas evitáveis e como esse comportamento evolui ao longo do período de 2015 a 2024.

A análise também mostra que a interpretação dos dados deve considerar a limitação da base, já que não há variável regional disponível. Ainda assim, o estudo permite apontar grupos prioritários para ações de prevenção e políticas públicas.

**Resposta à pergunta inicial:**
