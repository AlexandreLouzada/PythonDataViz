
# Projeto 16 — Gastos Públicos por Município e Indicadores Sociais (2015–2024)

**Objetivo:** Analisar a distribuição dos gastos públicos por município e sua relação com indicadores sociais.

---



## 1. Problema

Como os gastos públicos estão distribuídos entre municípios e existe relação com indicadores sociais?

---



## 2. KPI

- Gasto público per capita  
- Indicador social (ex: IDH, educação, saúde)

👉 Objetivo: avaliar eficiência e desigualdade.


```

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_excel("Simulacao_Gastos_Publicos_Municipios_2015_2024.xlsx")
df.head()
```

```

df.info()
df.isnull().sum()
```

```

df = df.rename(columns={
    "Gasto Publico": "Gasto",
    "Populacao": "Populacao",
    "Indicador Social": "Indicador"
})

df["Gasto_per_capita"] = df["Gasto"] / df["Populacao"]
df.head()
```

```

# Média por município
media_mun = df.groupby("Municipio")["Gasto_per_capita"].mean().sort_values(ascending=False)
media_mun
```

```

# Correlação
df[["Gasto_per_capita","Indicador"]].corr()
```

```

sns.scatterplot(data=df, x="Gasto_per_capita", y="Indicador")
plt.title("Gasto Público vs Indicador Social")
plt.show()
```


## Interpretação

✔ Correlação positiva:
> Mais gasto pode estar associado a melhores indicadores.

✔ Correlação fraca:
> Gasto não garante eficiência.
