## 1. Introdução

As queimadas na Amazônia Legal representam um problema ambiental de grande impacto ecológico, climático e social. A análise histórica desses dados ajuda a identificar padrões, estados mais afetados e anos de maior criticidade, apoiando ações de monitoramento e políticas públicas ambientais.

Neste projeto, vamos investigar a evolução dos focos de queimadas na Amazônia Legal ao longo dos últimos 10 anos.



## 2. Definição do problema

**Pergunta de análise:**  
Como os focos de queimadas evoluíram ao longo dos últimos anos na Amazônia Legal e quais estados apresentaram maior ocorrência?

**Por que essa pergunta é relevante?**  
Porque ajuda a identificar padrões temporais e espaciais do problema, apoiando ações de monitoramento ambiental, prevenção e formulação de políticas públicas.

**Qual decisão poderia ser tomada com base nessa análise?**  
Priorizar ações de fiscalização, prevenção e combate em estados e períodos com maior incidência de queimadas.



## 3. Definição do KPI

### KPI principal
**Total de focos de queimadas por estado e por ano**

### Justificativa
Esse indicador permite comparar:
- a evolução do problema ao longo do tempo;
- os estados mais afetados;
- os anos mais críticos.

### Fórmula
\[
	ext{Total de Focos} = \sum 	ext{Focos de Queimadas}
\]

### KPI complementar
**Variação percentual anual por estado**

Esse indicador ajuda a identificar crescimento, redução ou picos anormais em determinados períodos.


```

# 4. Aquisição dos dados

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["figure.figsize"] = (10, 6)
sns.set_theme(style="whitegrid")

# Leitura da base
df = pd.read_excel("Simulacao_Queimadas_Amazonia_Legal_2014_2023.xlsx")

# Visualização inicial
df.head()
```


### Descrição das variáveis principais
- **Ano**: ano de referência
- **Estado**: unidade da federação da Amazônia Legal
- **Focos de Queimadas**: quantidade registrada de focos no período


```

# 5. Limpeza e preparação dos dados

print("Dimensões da base:", df.shape)
print("\nTipos de dados:")
print(df.dtypes)

print("\nValores ausentes por coluna:")
print(df.isnull().sum())

print("\nQuantidade de linhas duplicadas:", df.duplicated().sum())
```

```

# Renomeando colunas para facilitar a escrita dos comandos
df = df.rename(columns={"Focos de Queimadas": "Focos_Queimadas"})

# Garantindo tipos corretos
df["Ano"] = df["Ano"].astype(int)
df["Estado"] = df["Estado"].astype(str)
df["Focos_Queimadas"] = df["Focos_Queimadas"].astype(int)

df.head()
```


### Comentário sobre a preparação
Nesta etapa, verificamos:
- estrutura da base;
- tipos de dados;
- valores ausentes;
- duplicidades.

Também renomeamos a coluna principal para facilitar os comandos em Python.


```

# 6. Análise exploratória

# Estatísticas descritivas
df["Focos_Queimadas"].describe()
```

```

# Total de focos por ano
focos_por_ano = df.groupby("Ano")["Focos_Queimadas"].sum().reset_index()
focos_por_ano
```

```

# Total de focos por estado
focos_por_estado = (
    df.groupby("Estado")["Focos_Queimadas"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)
focos_por_estado
```

```

# Tabela dinâmica: focos por ano e por estado
focos_estado_ano = df.pivot_table(
    index="Ano",
    columns="Estado",
    values="Focos_Queimadas",
    aggfunc="sum"
)

focos_estado_ano
```

```

# Variação percentual anual por estado
df = df.sort_values(["Estado", "Ano"])
df["Variacao_Percentual"] = df.groupby("Estado")["Focos_Queimadas"].pct_change() * 100

df.head(10)
```


## 7. Visualizações

Os gráficos abaixo ajudam a responder:
- como o problema evolui ao longo do tempo;
- quais estados concentram mais focos;
- em quais anos houve maior criticidade.


```

# Gráfico 1: evolução total dos focos ao longo do tempo
plt.figure(figsize=(10, 6))
sns.lineplot(data=focos_por_ano, x="Ano", y="Focos_Queimadas", marker="o")
plt.title("Evolução Total dos Focos de Queimadas na Amazônia Legal (2014–2023)")
plt.xlabel("Ano")
plt.ylabel("Focos de Queimadas")
plt.xticks(focos_por_ano["Ano"], rotation=45)
plt.show()
```

```

# Gráfico 2: total de focos por estado
plt.figure(figsize=(10, 6))
sns.barplot(data=focos_por_estado, x="Estado", y="Focos_Queimadas")
plt.title("Total de Focos de Queimadas por Estado")
plt.xlabel("Estado")
plt.ylabel("Focos de Queimadas")
plt.xticks(rotation=45)
plt.show()
```

```

# Gráfico 3: evolução dos focos por estado
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x="Ano", y="Focos_Queimadas", hue="Estado", marker="o")
plt.title("Evolução dos Focos de Queimadas por Estado")
plt.xlabel("Ano")
plt.ylabel("Focos de Queimadas")
plt.xticks(sorted(df["Ano"].unique()), rotation=45)
plt.legend(title="Estado", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```

# Gráfico 4: distribuição dos focos de queimadas
plt.figure(figsize=(8, 5))
sns.histplot(df["Focos_Queimadas"], bins=10, kde=True)
plt.title("Distribuição dos Focos de Queimadas")
plt.xlabel("Focos de Queimadas")
plt.ylabel("Frequência")
plt.show()
```


## 8. Insights e interpretação

Nesta etapa, o foco é explicar os padrões observados, e não apenas descrever os gráficos.


```

# Estado com maior total de focos
estado_mais_critico = focos_por_estado.loc[focos_por_estado["Focos_Queimadas"].idxmax()]
estado_mais_critico
```

```

# Ano com maior número total de focos
ano_mais_critico = focos_por_ano.loc[focos_por_ano["Focos_Queimadas"].idxmax()]
ano_mais_critico
```


### Modelo de interpretação esperada

**Insight 1 — Evolução temporal**  
A série histórica mostra que os focos de queimadas não se mantêm constantes ao longo do tempo. Há anos com crescimento mais intenso, indicando períodos críticos que podem estar relacionados a fatores climáticos, pressão por uso da terra e intensidade da fiscalização ambiental.

**Insight 2 — Diferenças entre estados**  
A comparação entre os estados mostra que o problema não afeta a Amazônia Legal de forma homogênea. Alguns estados concentram maior número de focos, o que pode refletir diferenças de extensão territorial, atividade agropecuária, desmatamento e capacidade de controle ambiental.

**Insight 3 — Anos críticos**  
Os anos com maior quantidade total de focos devem ser observados com atenção, pois podem indicar momentos de agravamento ambiental e necessidade de reforço em políticas de prevenção e fiscalização.

**Insight 4 — Limitações da análise**  
Embora a base permita identificar padrões por estado e por ano, ela não inclui variáveis como chuva, temperatura, desmatamento, fiscalização ou uso do solo. Portanto, os resultados mostram o padrão do problema, mas não explicam sozinhos todas as causas.



## 9. Conclusão

Com base na análise realizada, é possível identificar que os focos de queimadas variam significativamente ao longo do tempo e entre os estados da Amazônia Legal. Há estados com maior concentração do problema e anos claramente mais críticos.

Esses resultados reforçam a necessidade de ações públicas contínuas de monitoramento, prevenção e fiscalização ambiental, especialmente nos estados e períodos com maior incidência.

Os focos de queimadas apresentaram oscilações relevantes entre 2014 e 2023, com concentração maior em determinados estados e anos críticos, indicando a necessidade de estratégias regionais mais direcionadas.
