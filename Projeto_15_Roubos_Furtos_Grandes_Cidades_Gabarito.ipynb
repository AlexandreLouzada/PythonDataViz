## 1. Introdução

Roubos e furtos de veículos impactam diretamente a segurança urbana, o patrimônio da população e os custos sociais das grandes cidades. A análise desses dados permite identificar locais e períodos mais críticos, auxiliando ações de prevenção e segurança pública.

Neste projeto, vamos investigar o comportamento dos roubos e furtos de veículos nas grandes cidades entre 2018 e 2024.



## 2. Definição do problema

**Pergunta de análise:**  
Como os roubos e furtos de veículos evoluíram nas grandes cidades entre 2018 e 2024 e quais localidades ou períodos apresentaram maior concentração de ocorrências?

**Por que essa pergunta é relevante?**  
Porque ajuda a identificar áreas e períodos críticos, apoiando decisões de policiamento, prevenção e gestão da segurança urbana.

**Qual decisão poderia ser tomada com base nessa análise?**  
Priorizar cidades ou períodos com maior número de ocorrências em estratégias de monitoramento, policiamento e prevenção.



## 3. Definição do KPI

### KPI principal
**Total de ocorrências de roubos e furtos de veículos**

### Justificativa
Esse indicador permite comparar:
- cidades mais afetadas;
- evolução temporal do problema;
- períodos com maior concentração de ocorrências.

### KPI complementar
- Média anual de ocorrências por cidade
- Variação percentual ao longo do tempo
- Ranking de cidades mais críticas


```

# 4. Aquisição dos dados

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["figure.figsize"] = (10, 6)
sns.set_theme(style="whitegrid")

# Leitura da base
df = pd.read_excel("Simulacao_Roubos_Furtos_Veiculos_Cidades_2018_2024.xlsx")

# Visualização inicial
df.head()
```


### Descrição das variáveis principais
- **Ano**: ano de referência
- **Cidade**: cidade analisada
- **Ocorrencias**: total de roubos e furtos de veículos no período
- Outras variáveis temporais (como mês, dia da semana ou horário), se estiverem disponíveis na base


```

# 5. Limpeza e preparação dos dados

print("Dimensões da base:", df.shape)
print("\nTipos de dados:")
print(df.dtypes)

print("\nValores ausentes por coluna:")
print(df.isnull().sum())

print("\nQuantidade de linhas duplicadas:", df.duplicated().sum())
```

```

# Ajuste defensivo de nomes de colunas
rename_map = {
    "Ocorrências": "Ocorrencias",
    "Cidade": "Cidade",
    "Ano": "Ano"
}
df = df.rename(columns={k: v for k, v in rename_map.items() if k in df.columns})

# Garantindo tipos corretos
df["Ano"] = df["Ano"].astype(int)
df["Cidade"] = df["Cidade"].astype(str)
df["Ocorrencias"] = df["Ocorrencias"].astype(int)

df.head()
```


### Comentário sobre a preparação
Nesta etapa:
- verificamos a estrutura da base;
- ajustamos nomes de colunas, quando necessário;
- garantimos o tipo correto das variáveis;
- preparamos os dados para comparações temporais e entre cidades.


```

# 6. Análise exploratória

# Estatísticas descritivas das ocorrências
df["Ocorrencias"].describe()
```

```

# Total de ocorrências por ano
ocorrencias_por_ano = df.groupby("Ano")["Ocorrencias"].sum().reset_index()
ocorrencias_por_ano
```

```

# Total de ocorrências por cidade
ocorrencias_por_cidade = (
    df.groupby("Cidade")["Ocorrencias"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)
ocorrencias_por_cidade
```

```

# Tabela dinâmica: ocorrências por ano e cidade
ocorrencias_cidade_ano = df.pivot_table(
    index="Ano",
    columns="Cidade",
    values="Ocorrencias",
    aggfunc="sum"
)

ocorrencias_cidade_ano
```

```

# Variação percentual por cidade ao longo do tempo
df = df.sort_values(["Cidade", "Ano"])
df["Variacao_Percentual"] = df.groupby("Cidade")["Ocorrencias"].pct_change() * 100

df.head(10)
```


## 7. Visualizações

Os gráficos abaixo ajudam a responder:
- quais cidades concentram mais ocorrências;
- como o problema evolui ao longo do tempo;
- em quais períodos há maior criticidade.


```

# Gráfico 1: evolução total das ocorrências ao longo do tempo
plt.figure(figsize=(10, 6))
sns.lineplot(data=ocorrencias_por_ano, x="Ano", y="Ocorrencias", marker="o")
plt.title("Evolução dos Roubos e Furtos de Veículos (2018–2024)")
plt.xlabel("Ano")
plt.ylabel("Ocorrências")
plt.xticks(ocorrencias_por_ano["Ano"], rotation=45)
plt.show()
```

```

# Gráfico 2: ranking das cidades com mais ocorrências
plt.figure(figsize=(10, 6))
sns.barplot(data=ocorrencias_por_cidade, x="Cidade", y="Ocorrencias")
plt.title("Total de Roubos e Furtos de Veículos por Cidade")
plt.xlabel("Cidade")
plt.ylabel("Ocorrências")
plt.xticks(rotation=45)
plt.show()
```

```

# Gráfico 3: evolução das ocorrências por cidade
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x="Ano", y="Ocorrencias", hue="Cidade", marker="o")
plt.title("Evolução dos Roubos e Furtos de Veículos por Cidade")
plt.xlabel("Ano")
plt.ylabel("Ocorrências")
plt.xticks(sorted(df["Ano"].unique()), rotation=45)
plt.legend(title="Cidade", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```

# Gráfico 4: distribuição das ocorrências
plt.figure(figsize=(8, 5))
sns.histplot(df["Ocorrencias"], bins=12, kde=True)
plt.title("Distribuição das Ocorrências de Roubos e Furtos")
plt.xlabel("Ocorrências")
plt.ylabel("Frequência")
plt.show()
```


## 8. Insights e interpretação

Nesta etapa, a análise deve explicar o que os dados revelam sobre risco, cidades críticas e tendência do problema.


```

# Cidade com maior total de ocorrências
cidade_mais_critica = ocorrencias_por_cidade.loc[ocorrencias_por_cidade["Ocorrencias"].idxmax()]
cidade_mais_critica
```

```

# Cidade com menor total de ocorrências
cidade_menos_critica = ocorrencias_por_cidade.loc[ocorrencias_por_cidade["Ocorrencias"].idxmin()]
cidade_menos_critica
```

```

# Ano com maior total de ocorrências
ano_mais_critico = ocorrencias_por_ano.loc[ocorrencias_por_ano["Ocorrencias"].idxmax()]
ano_mais_critico
```


### Modelo de interpretação esperada

**Insight 1 — Diferenças entre cidades**  
A comparação entre as cidades mostra que o problema não se distribui de forma homogênea. Algumas localidades concentram maior número de ocorrências, o que pode refletir fatores como tamanho da frota, densidade urbana e dinâmica da criminalidade.

**Insight 2 — Evolução temporal**  
A série histórica permite observar se o número de ocorrências está aumentando, diminuindo ou se mantém estável ao longo do tempo. Isso ajuda a perceber se houve agravamento ou melhora no cenário de segurança.

**Insight 3 — Períodos críticos**  
Os anos com maior número total de ocorrências merecem atenção especial, pois podem indicar momentos de maior vulnerabilidade ou mudanças no contexto de segurança pública.

**Insight 4 — Limitações da análise**  
A base mostra padrões por cidade e por ano, mas não explica sozinha as causas das variações. Uma análise mais aprofundada exigiria dados sobre frota, renda, policiamento, localização e outros fatores urbanos.
