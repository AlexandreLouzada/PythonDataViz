
# Projeto 5 — Computadores por Aluno vs Indicadores Educacionais

## Objetivo
Analisar se a disponibilidade de computadores por aluno nas escolas públicas está relacionada com indicadores educacionais.

---



## 1. Problema

Existe relação entre o número de computadores por aluno e o desempenho educacional nas escolas públicas?

---



## 2. KPI

### KPI principal:
- Computadores por aluno

### KPI complementar:
- Indicador educacional (ex: IDEB)

👉 Objetivo: verificar relação entre infraestrutura tecnológica e desempenho.


```

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_excel("Simulacao_Computadores_por_Aluno_Escolas_Publicas_2015_2024.xlsx")
df.head()
```

```

df.info()
df.isnull().sum()
```

```

# Ajuste de colunas (exemplo esperado)
df = df.rename(columns={
    "Computadores por Aluno": "Comp_Aluno",
    "Indicador Educacional": "Indicador"
})

df.head()
```

```

# Evolução temporal média
media_ano = df.groupby("Ano")[["Comp_Aluno","Indicador"]].mean()
media_ano
```

```

# Correlação
correlacao = df[["Comp_Aluno","Indicador"]].corr()
correlacao
```

```

# Gráfico temporal
media_ano.plot(title="Computadores por Aluno vs Indicador Educacional")
plt.show()
```

```

# Scatter plot (ESSENCIAL)
sns.scatterplot(data=df, x="Comp_Aluno", y="Indicador")
plt.title("Relação entre Computadores por Aluno e Desempenho")
plt.show()
```


## 6. Interpretação

### Cenário esperado:

✔ Correlação positiva:
> Quanto maior o número de computadores por aluno, melhor tende a ser o indicador educacional.

✔ Correlação fraca:
> A tecnologia sozinha não explica o desempenho educacional.
