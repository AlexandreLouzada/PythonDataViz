
# Projeto 9 — Relação entre Volume de Chuvas e Deslizamentos no RJ (2015–2023)

## Objetivo
Analisar se existe relação entre o volume de chuvas e a ocorrência de deslizamentos no estado do Rio de Janeiro.

---



## 1. Problema

Existe relação entre o volume de chuvas e o número de deslizamentos registrados ao longo do tempo?

---



## 2. KPI

### KPI principal:
- Volume de chuva (mm)
- Número de deslizamentos

👉 Objetivo: verificar se há relação entre as variáveis.


```

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_excel("Simulacao_Chuva_Deslizamentos_RJ_2015_2023.xlsx")
df.head()
```

```

df.info()
df.isnull().sum()
```

```

# Ajuste de colunas
df = df.rename(columns={
    "Volume de Chuva": "Chuva",
    "Deslizamentos": "Deslizamentos"
})

df.head()
```

```

# Evolução temporal
media_ano = df.groupby("Ano")[["Chuva","Deslizamentos"]].mean()
media_ano
```

```

# Correlação
correlacao = df[["Chuva","Deslizamentos"]].corr()
correlacao
```

```

# Gráfico temporal
media_ano.plot(title="Chuva vs Deslizamentos ao longo do tempo")
plt.show()
```

```

# Scatter plot (ESSENCIAL)
sns.scatterplot(data=df, x="Chuva", y="Deslizamentos")
plt.title("Relação entre Chuva e Deslizamentos")
plt.show()
```


## 6. Interpretação

### Cenário esperado:

✔ Correlação positiva:
> Maior volume de chuva tende a aumentar o número de deslizamentos.

✔ Correlação fraca:
> Outros fatores também influenciam (ocupação do solo, infraestrutura, relevo).
