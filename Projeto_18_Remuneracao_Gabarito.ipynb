```

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_excel("Simulacao_Remuneracao_Servidores_2015_2024.xlsx")
df.head()
```

```

df.info()
df.isnull().sum()
```

```

df = df.rename(columns={
    "Remuneracao": "Remuneracao",
    "Cargo": "Cargo"
})

df.head()
```

```

# Média por cargo
media_cargo = df.groupby("Cargo")["Remuneracao"].mean().sort_values(ascending=False)
media_cargo
```

```

# Estatísticas gerais
df["Remuneracao"].describe()
```

```

# Evolução temporal
media_ano = df.groupby("Ano")["Remuneracao"].mean()
media_ano
```

```

# Gráfico evolução
media_ano.plot(title="Evolução da Remuneração Média")
plt.show()
```

```

# Boxplot (ESSENCIAL)
sns.boxplot(data=df, x="Cargo", y="Remuneracao")
plt.title("Distribuição Salarial por Cargo")
plt.xticks(rotation=45)
plt.show()
```


## 8. Insights

✔ Diferença entre cargos:
> Alguns cargos apresentam remuneração significativamente maior.

✔ Distribuição:
> A presença de outliers pode indicar distorções salariais.
