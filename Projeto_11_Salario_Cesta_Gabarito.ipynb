
# Projeto 11 — Evolução do Salário Mínimo vs Custo da Cesta Básica (2010–2024)

## Objetivo
Analisar o poder de compra da população ao longo do tempo, comparando o salário mínimo com o custo da cesta básica.

---



## 1. Problema

Como evoluiu o poder de compra da população brasileira ao longo do tempo considerando o salário mínimo e o custo da cesta básica?

---



## 2. KPI

### KPI principal:
- Quantidade de cestas básicas que podem ser compradas com 1 salário mínimo

### Fórmula:
Cestas Compradas = Salário Mínimo / Custo da Cesta

👉 Esse KPI mede diretamente o poder de compra.

---


```

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_excel("Simulacao_Salario_Cesta_Estados_2010_2024.xlsx")
df.head()
```

```

df.info()
df.isnull().sum()
```

```

# Ajuste de colunas
df = df.rename(columns={
    "Salario Minimo": "Salario",
    "Custo Cesta": "Cesta"
})

df.head()
```

```

# KPI: poder de compra
df["Cestas_Compradas"] = df["Salario"] / df["Cesta"]

df.head()
```

```

# Evolução temporal média
media_ano = df.groupby("Ano")[["Salario","Cesta","Cestas_Compradas"]].mean()
media_ano
```

```

# Gráfico evolução
media_ano["Cestas_Compradas"].plot(title="Evolução do Poder de Compra")
plt.ylabel("Cestas por salário")
plt.show()
```

```

# Identificação de extremos
melhor_ano = media_ano["Cestas_Compradas"].idxmax()
pior_ano = media_ano["Cestas_Compradas"].idxmin()

melhor_ano, pior_ano
```


## 6. Interpretação

### Modelo esperado:

> O poder de compra varia ao longo do tempo e depende da relação entre salário e custo de vida.

> Mesmo com aumento do salário, o poder de compra pode cair se o custo da cesta subir mais rapidamente.

---



## 7. Conclusão

A análise permite entender se o aumento do salário mínimo realmente melhora a vida da população.
