
# Projeto 20 — Acesso à Internet nos Municípios (2010–2024)

**Objetivo:** Analisar a evolução do acesso à internet nos municípios brasileiros, identificando desigualdades e tendências ao longo do tempo.

---



## 1. Problema

Como o acesso à internet evoluiu nos municípios ao longo do tempo e existem desigualdades significativas entre eles?

---



## 2. KPI

- Percentual de domicílios com acesso à internet  
- Crescimento ao longo do tempo  

👉 Objetivo: identificar inclusão digital e desigualdade


```

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_excel("Simulacao_Acesso_Internet_Municipios_2010_2024.xlsx")
df.head()
```

```

df.info()
df.isnull().sum()
```

```

df = df.rename(columns={
    "Acesso Internet (%)": "Internet",
    "Municipio": "Municipio"
})

df.head()
```

```

# Média por município
media_municipio = df.groupby("Municipio")["Internet"].mean().sort_values(ascending=False)
media_municipio
```

```

# Evolução temporal
media_ano = df.groupby("Ano")["Internet"].mean()
media_ano
```

```

# Gráfico evolução
media_ano.plot(title="Evolução do Acesso à Internet")
plt.ylabel("% acesso")
plt.show()
```

```

# Ranking municípios
media_municipio.head(10).plot(kind="bar", title="Municípios com maior acesso")
plt.show()
```

```

# Crescimento percentual
df = df.sort_values(["Municipio","Ano"])
df["Crescimento"] = df.groupby("Municipio")["Internet"].pct_change()*100
df.head()
```


## Interpretação

✔ Crescimento:
> O acesso à internet tende a crescer ao longo do tempo, indicando avanço da inclusão digital.

✔ Desigualdade:
> Alguns municípios apresentam níveis significativamente mais baixos, indicando exclusão digital.
