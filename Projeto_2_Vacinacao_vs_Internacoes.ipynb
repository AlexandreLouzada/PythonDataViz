# Projeto 2 — Vacinação COVID-19 vs Internações


## 1. Problema
Como a taxa de vacinação contra COVID-19 se relaciona com as internações nos municípios ao longo do tempo?

## 2. KPI
- Taxa de vacinação (%)
- Taxa de internação (% ou por população)

Objetivo: verificar se existe relação entre aumento da vacinação e redução das internações.


```

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel("Simulacao_Vacinacao_COVID_Municipios_2021_2024.xlsx")
df.head()
```

```

df.info()
df.isnull().sum()
```

```

# Ajuste de nomes (exemplo esperado)
df = df.rename(columns={
    "Taxa Vacinação": "Taxa_Vacinacao",
    "Taxa Internações": "Taxa_Internacao"
})

df.head()
```

```

# Evolução temporal média
media_ano = df.groupby("Ano")[["Taxa_Vacinacao","Taxa_Internacao"]].mean()
media_ano
```

```

# Correlação
correlacao = df[["Taxa_Vacinacao","Taxa_Internacao"]].corr()
correlacao
```

```

# Gráfico temporal
media_ano.plot(title="Vacinação vs Internações ao longo do tempo")
plt.show()
```

```

# Scatter plot (ESSENCIAL)
sns.scatterplot(data=df, x="Taxa_Vacinacao", y="Taxa_Internacao")
plt.title("Relação entre Vacinação e Internações")
plt.show()
```


## 6. Interpretação (Modelo)

Se houver correlação negativa:

> Observa-se que, à medida que a taxa de vacinação aumenta, a taxa de internações tende a diminuir, sugerindo um efeito protetor da vacinação.

Se não houver correlação forte:

> A relação entre vacinação e internações não é claramente linear, indicando que outros fatores podem influenciar as internações.




## 7. Conclusão

A análise permite avaliar se a vacinação impacta diretamente nas internações. Em geral, espera-se uma relação inversa, mas fatores externos também devem ser considerados.
