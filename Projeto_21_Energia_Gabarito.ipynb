
# Projeto 21 — Produção de Energia por Fonte no Brasil (2010–2024)

**Objetivo:** Analisar a evolução da produção de energia por diferentes fontes, identificando mudanças na matriz energética brasileira.

---



## 1. Problema

Como a produção de energia evoluiu ao longo do tempo e quais fontes estão crescendo ou diminuindo na matriz energética?

---



## 2. KPI

- Produção total por fonte  
- Participação percentual por fonte  

👉 Objetivo: identificar mudanças na matriz energética


```

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_excel("Simulacao_Producao_Energia_Brasil_2010_2024.xlsx")
df.head()
```

```

df.info()
df.isnull().sum()
```

```

df = df.rename(columns={
    "Fonte Energia": "Fonte",
    "Producao": "Producao"
})

df.head()
```

```

# Total por fonte
total_fonte = df.groupby("Fonte")["Producao"].sum().sort_values(ascending=False)
total_fonte
```

```

# Evolução temporal
total_ano = df.groupby("Ano")["Producao"].sum()
total_ano
```

```

# Evolução por fonte
plt.figure(figsize=(12,6))
sns.lineplot(data=df, x="Ano", y="Producao", hue="Fonte")
plt.title("Evolução da Produção de Energia por Fonte")
plt.show()
```

```

# Participação percentual
total_ano_fonte = df.groupby(["Ano","Fonte"])["Producao"].sum().reset_index()

total_ano_fonte["Percentual"] = total_ano_fonte.groupby("Ano")["Producao"].transform(lambda x: x / x.sum() * 100)

total_ano_fonte.head()
```

```

# Gráfico de participação
pivot = total_ano_fonte.pivot(index="Ano", columns="Fonte", values="Percentual")

pivot.plot.area(title="Participação das Fontes de Energia")
plt.ylabel("% participação")
plt.show()
```


## Interpretação

✔ Crescimento:
> Algumas fontes renováveis podem estar aumentando sua participação.

✔ Mudança de matriz:
> Redução de fontes tradicionais pode indicar transição energética.
