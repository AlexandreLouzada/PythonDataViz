## 1. Introdução

O ENEM é um dos principais indicadores de desempenho educacional no Brasil. A análise dos resultados por região permite identificar desigualdades, tendências e possíveis diferenças estruturais de acesso, qualidade do ensino e oportunidades educacionais.

Neste projeto, vamos investigar como o desempenho no ENEM evoluiu entre 2015 e 2023 nas diferentes regiões do país.



## 2. Definição do problema

**Pergunta de análise:**  
Como os resultados do ENEM evoluíram entre 2015 e 2023 nas regiões do Brasil e quais regiões apresentaram os melhores e piores desempenhos?

**Por que essa pergunta é relevante?**  
Porque ajuda a identificar desigualdades regionais no desempenho educacional e pode apoiar políticas públicas voltadas à melhoria da qualidade do ensino.

**Qual decisão poderia ser tomada com base nessa análise?**  
Direcionar programas educacionais e investimentos para regiões com menor desempenho ou maior defasagem.



## 3. Definição do KPI

### KPI principal
**Média das notas do ENEM por região**

### Justificativa
Esse indicador permite comparar o desempenho médio das regiões ao longo do tempo.

### Fórmula
\[
	ext{Média Regional} = rac{\sum 	ext{Notas}}{	ext{Quantidade de Registros}}
\]

### KPI complementar
**Média por área do ENEM e por região**

Esse indicador ajuda a verificar se determinadas regiões apresentam melhor desempenho em áreas específicas, como Matemática ou Linguagens.


```

# 4. Aquisição dos dados

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["figure.figsize"] = (10, 6)
sns.set_theme(style="whitegrid")

# Leitura da base
df = pd.read_excel("Simulacao_Resultados_ENEM_Regioes_2015_2023.xlsx")

# Visualização inicial
df.head()
```


### Descrição das variáveis principais
- **Ano**: ano de referência da nota
- **Região**: região do Brasil
- **Área**: área de conhecimento avaliada no ENEM
- **Nota Média**: média obtida naquela combinação de ano, região e área


```

# 5. Limpeza e preparação dos dados

print("Dimensões da base:", df.shape)
print("\nTipos de dados:")
print(df.dtypes)

print("\nValores ausentes por coluna:")
print(df.isnull().sum())

print("\nQuantidade de linhas duplicadas:", df.duplicated().sum())
```

```

# Renomeando colunas para facilitar os comandos
df = df.rename(columns={
    "Nota Média": "Nota_Media",
    "Região": "Regiao",
    "Área": "Area"
})

# Garantindo tipos corretos
df["Ano"] = df["Ano"].astype(int)
df["Regiao"] = df["Regiao"].astype(str)
df["Area"] = df["Area"].astype(str)
df["Nota_Media"] = df["Nota_Media"].astype(float)

df.head()
```


### Comentário sobre a preparação
Nesta etapa, verificamos:
- estrutura da base;
- tipos de dados;
- valores ausentes;
- duplicidades.

Também renomeamos algumas colunas para simplificar os comandos em Python.


```

# 6. Análise exploratória

# Estatísticas descritivas da nota média
df["Nota_Media"].describe()
```

```

# Média da nota por região
media_por_regiao = (
    df.groupby("Regiao")["Nota_Media"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)
media_por_regiao
```

```

# Média da nota por ano
media_por_ano = df.groupby("Ano")["Nota_Media"].mean().reset_index()
media_por_ano
```

```

# Média da nota por ano e por região
media_regiao_ano = df.pivot_table(
    index="Ano",
    columns="Regiao",
    values="Nota_Media",
    aggfunc="mean"
)

media_regiao_ano
```

```

# Média por área e por região
media_area_regiao = df.pivot_table(
    index="Area",
    columns="Regiao",
    values="Nota_Media",
    aggfunc="mean"
)

media_area_regiao
```

```

# Variação percentual da média por região ao longo do tempo
df = df.sort_values(["Regiao", "Ano"])
df["Variacao_Percentual"] = df.groupby("Regiao")["Nota_Media"].pct_change() * 100

df.head(10)
```


## 7. Visualizações
Os gráficos abaixo ajudam a responder:
- quais regiões têm melhor desempenho médio;
- como o desempenho evolui ao longo do tempo;
- como as regiões se comportam em diferentes áreas do ENEM.


```

# Gráfico 1: média geral por região
plt.figure(figsize=(8, 5))
sns.barplot(data=media_por_regiao, x="Regiao", y="Nota_Media")
plt.title("Média das Notas do ENEM por Região")
plt.xlabel("Região")
plt.ylabel("Nota Média")
plt.show()
```

```

# Gráfico 2: evolução da nota média geral ao longo do tempo
plt.figure(figsize=(10, 6))
sns.lineplot(data=media_por_ano, x="Ano", y="Nota_Media", marker="o")
plt.title("Evolução da Nota Média do ENEM no Brasil (2015–2023)")
plt.xlabel("Ano")
plt.ylabel("Nota Média")
plt.xticks(media_por_ano["Ano"], rotation=45)
plt.show()
```

```

# Gráfico 3: evolução da nota média por região
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x="Ano", y="Nota_Media", hue="Regiao", marker="o")
plt.title("Evolução da Nota Média do ENEM por Região")
plt.xlabel("Ano")
plt.ylabel("Nota Média")
plt.xticks(sorted(df["Ano"].unique()), rotation=45)
plt.legend(title="Região")
plt.show()
```

```

# Gráfico 4: heatmap das médias por área e região
plt.figure(figsize=(10, 6))
sns.heatmap(media_area_regiao, annot=True, fmt=".1f", cmap="Blues")
plt.title("Nota Média por Área do ENEM e Região")
plt.xlabel("Região")
plt.ylabel("Área")
plt.show()
```


## 8. Insights e interpretação

Nesta etapa, é importante ir além da descrição dos gráficos. O foco deve estar em explicar desigualdades, tendências e possíveis implicações.


```

# Região com maior média geral
regiao_melhor = media_por_regiao.loc[media_por_regiao["Nota_Media"].idxmax()]
regiao_melhor
```

```

# Região com menor média geral
regiao_pior = media_por_regiao.loc[media_por_regiao["Nota_Media"].idxmin()]
regiao_pior
```


### Modelo de interpretação esperada

**Insight 1 — Desigualdade regional**  
A comparação entre regiões mostra que o desempenho no ENEM não é homogêneo no Brasil. Algumas regiões apresentam médias sistematicamente superiores, enquanto outras permanecem abaixo, o que pode refletir desigualdades históricas de acesso, infraestrutura e qualidade do ensino.

**Insight 2 — Evolução temporal**  
Ao observar a evolução ao longo do tempo, é possível identificar se houve melhora, estabilidade ou queda nas notas médias. Isso ajuda a perceber se a desigualdade está diminuindo, aumentando ou se mantém constante.

**Insight 3 — Diferenças por área**  
As áreas do ENEM podem revelar padrões distintos. Uma região pode se destacar mais em uma área específica e apresentar dificuldades em outra, o que ajuda a refinar a interpretação educacional.

**Insight 4 — Limitações**  
A base mostra médias por região e área, mas não explica sozinha as causas das desigualdades. Para uma análise mais profunda, seria importante incluir dados sobre renda, infraestrutura escolar, acesso à internet e formação docente.
